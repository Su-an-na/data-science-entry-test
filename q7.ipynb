### Task 1

````python
class Car:
    def __init__(self, make, model, year):
        self.make = make
        self.model = model
        self.year = year
        
    def describe_car(self):
        print(f"Year: {self.year}, Make: {self.make}, Model: {self.model}")
````

### Task 2

````python
c1 = Car("Toyota", "Corolla", 2020)
c1.describe_car()
````
